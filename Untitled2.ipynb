```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
```

```python
df_leido = pd.read_csv("dataset_ciberseguridad_500.csv")
df_leido.columns = df_leido.columns.str.strip()
```

```python
col_y = "data_transfer_mb"
col_x = "request_frequency_per_min"
for col in [col_x, col_y]:
    if df_leido[col].dtype == object:
        df_leido[col] = df_leido[col].astype(str).str.replace(r'[^0-9.\-]', '', regex=True)
    df_leido[col] = pd.to_numeric(df_leido[col], errors='coerce')
```

```python
df_leido = df_leido.dropna(subset=[col_x, col_y])

df_leido = df_leido[
    (df_leido[col_x] < df_leido[col_x].quantile(0.99)) & 
    (df_leido[col_y] < df_leido[col_y].quantile(0.99))
]
```

```python
X = df_leido[[col_x, col_y]].values
mean = np.mean(X, axis=0)
std = np.std(X, axis=0)
X_scaled = (X - mean) / std
```

#### optimal_k = 3
kmeans = KMeans(n_clusters=optimal_k, random_state=42, n_init=10)
kmeans.fit(X_scaled)

```python
labels = kmeans.labels_
centroids_scaled = kmeans.cluster_centers_
centroids = centroids_scaled * std + mean
df_leido["Cluster"] = labels
```

```python
plt.figure(figsize=(9, 7))
scatter = plt.scatter(df_leido[col_x], df_leido[col_y], c=labels, cmap='viridis', edgecolors='black', s=65, alpha=0.85)
plt.scatter(centroids[:, 0], centroids[:, 1], c='red', marker='X', s=220, label='Centroides')
plt.title("Datos ciberseguridad", fontsize=13, fontweight='bold')
plt.xlabel("Frecuencia", fontsize=11)
plt.ylabel("MB", fontsize=11)
plt.grid(True, linestyle='-', linewidth=0.7, alpha=0.7)
plt.legend()
plt.show()
```
