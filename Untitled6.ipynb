```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
from sklearn.decomposition import PCA
```

```python
df = pd.read_csv('cafe_sucio.csv', na_values=['?'])

cols_num = ['Altitud_Cultivo', 'Peso_Grano_g', 'Porcentaje_Humedad', 'Puntaje_Catacion']

for col in cols_num:
    df[col] = pd.to_numeric(df[col], errors='coerce')
    df[col] = df[col].where(df[col] >= 0)
    df[col] = df[col].fillna(df[col].mean())

df.loc[df['Porcentaje_Humedad'] > 15, 'Porcentaje_Humedad'] = np.nan
df['Porcentaje_Humedad'] = df['Porcentaje_Humedad'].fillna(df['Porcentaje_Humedad'].mean())

df['Altitud_Cultivo'] = df['Altitud_Cultivo'].round(2)
df['Peso_Grano_g'] = df['Peso_Grano_g'].round(4)
df['Porcentaje_Humedad'] = df['Porcentaje_Humedad'].round(2)
df['Puntaje_Catacion'] = df['Puntaje_Catacion'].round(2)

df = df.drop(columns=['Color_Principal', 'ID_Lote'])

df.to_csv('datos_limpio2.csv', index=False, encoding='utf-8')        
```

```python

labels = kmeans.labels_
centroids = kmeans.cluster_centers_
```

```python

orden = np.argsort(centroids[:, 2])
nombres = {orden[0]: 'Malo', orden[1]: 'Bueno', orden[2]: 'Premium'}

df_leido["Cluster"] = labels
df_leido["Calidad"] = df_leido["Cluster"].map(nombres)
```

```python
col_x = 'Porcentaje_Humedad'
col_y = 'Puntaje_Catacion'

plt.figure(figsize=(9, 7))
plt.scatter(df_leido[col_x], df_leido[col_y], c=labels, cmap='viridis',
            edgecolors='black', s=65, alpha=0.85)
plt.scatter(centroids[:, 1], centroids[:, 2], c='red', marker='X', s=220, label='Centroides')
for i in range(3):
    plt.annotate(nombres[i], (centroids[i, 1], centroids[i, 2]),
                 textcoords="offset points", xytext=(8, 8), fontsize=10)
plt.title("Clusters de Café", fontsize=13, fontweight='bold')
plt.xlabel("Porcentaje de Humedad (%)", fontsize=11)
plt.ylabel("Puntaje de Catación", fontsize=11)
plt.grid(True, linestyle='-', linewidth=0.7, alpha=0.7)
plt.legend()
plt.show()
```
